```python
import random


class simpleLLM:
    def __init__(self):
        print("LLM Created")

    def predict(self, prompt):
        response_list = [
            'Delihi is capital of India',
            'The capital of India is Delhi',
            'Artificial intelligence + Cybersecurity is the future',
        ] 
        return {"response": random.choice(response_list)}  # Return a random  response for simplicity
```

```python
llm = simpleLLM()
response = llm.predict("What is the capital of India?")
print(response)
```

```python
class simplePromptTemplate:
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
 
    def format(self, **kwargs):
        return self.template.format(**kwargs)
```

```python
template = simplePromptTemplate(
    "What is the capital of {country}?",
    input_variables=["country"]
)
```

```python
prompt = template.format(country="India")
```

```python
llm = simpleLLM()
```

```python
llm.predict(prompt)
```

Now comes the chaining part

```python
class simpleLLMChain:
    def __init__(self, llm, prompt_template):
        self.llm = llm
        self.prompt_template = prompt_template

    def run(self, input_dict):
        prompt = self.prompt_template.format(**input_dict)
        res =  self.llm.predict(prompt) 
        return res['response']
    
```

```python
template1 = simplePromptTemplate(
    "What is the capital of {country}?",
    input_variables=["country"]
)
```

```python
llm = simpleLLM()
```

```python
chain1 = simpleLLMChain(llm, template1)
```

```python
response = chain1.run({'country': "India"})
print(response)
```
