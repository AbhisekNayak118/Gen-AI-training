```python
from abc import ABC, abstractmethod

class Runnable(ABC):
    @abstractmethod
    def invoke(self, input_data):
        pass
```

```python
import random

class simpleLLM(Runnable):
    def __init__(self):
        print("LLM Created")

    def invoke(self, prompt):
        response_list = [
            'Delihi is capital of India',
            'The capital of India is Delhi',
            'Artificial intelligence + Cybersecurity is the future',
        ] 
        
        return {"response": random.choice(response_list)}  # Return a random  response for simplicity

    def predict(self, prompt):
        response_list = [
            'Delihi is capital of India',
            'The capital of India is Delhi',
            'Artificial intelligence + Cybersecurity is the future',
        ] 
        print("This methods will be deprecated in future, we sugest you should use invoke method instead of predict.")
        return {"response": random.choice(response_list)}  # Return a random  response for simplicity
```

```python
llm = simpleLLM()
```

```python
class simplePromptTemplate(Runnable):
    def __init__(self, template, input_variables):
        self.template = template
        self.input_variables = input_variables
    def invoke(self, input_data):
        return self.template.format(**input_data)
    
    print("This methods will be deprecated in future, we sugest you should use invoke method instead of predict.")
    def format(self, **kwargs):
        return self.template.format(**kwargs)
```

```python
class simpleParser(Runnable):
    def __init__(self):
       pass

    def invoke(self, input_data):
        return input_data['response']
```

Now the point is to convert this 2 programs into runnables

```python
class connectorRunnable(Runnable):
    def __init__(self, runnable_list):
        self.runnable_list = runnable_list
        # self.prompt_template = prompt_template
    
    def invoke(self, input_data):
        for runnable in self.runnable_list:
            input_data = runnable.invoke(input_data)
        return input_data
```

```python
template = simplePromptTemplate(
    "What is the capital of {country}?",
    input_variables=["country"]
)
```

```python
llm = simpleLLM()
```

```python
parser = simpleParser()
```

```python
chain = connectorRunnable([template, llm, parser])
```

```python
chain_res = chain.invoke({'country': "India"})
print(chain_res)
```
