```python
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnableSequence, RunnableParallel, RunnablePassthrough, RunnableLambda

from dotenv import load_dotenv
import os

load_dotenv() 
```

```python
prompt1 = PromptTemplate(
    template = 'Generate a tweet about {topic}',
    input_variables = ['topic']
)
```

```python
llm = HuggingFaceEndpoint(
    model="MiniMaxAI/MiniMax-M2.5",
    task = "text-generation",
    huggingfacehub_api_token=os.getenv("HUGGINGFACEHUB_API_TOKEN")
)
```

```python
model = ChatHuggingFace(llm = llm)
```

```python
parser = StrOutputParser()
```

```python
joke_generator = RunnableSequence(prompt1, model, parser)
```

```python
def wordCount(text: str) -> int:
    return len(text.split())
```

```python

parallel_chain = RunnableParallel({
    'joke': RunnablePassthrough(),
    'wordCount': RunnableLambda(wordCount),

    # 'wordCount': RunnableLambda(lambda x: len(x.split())),

})
```

```python
chain = RunnableSequence(joke_generator, parallel_chain)
```

```python
final_chain = chain.invoke({"topic": "Artificial Intelligence"})
```

```python
final_res = """ {} \n word count - {}""".format(final_chain['joke'], final_chain['wordCount'])
```
