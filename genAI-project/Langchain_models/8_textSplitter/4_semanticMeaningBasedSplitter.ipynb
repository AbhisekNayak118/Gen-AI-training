```python
# from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_experimental.text_splitter import SemanticChunker

from langchain_huggingface.embeddings import HuggingFaceEmbeddings

from dotenv import load_dotenv
import os

load_dotenv()
```

```python
semantic_text_splitter = SemanticChunker(
        embeddings=HuggingFaceEmbeddings(),
        breakpoint_threshold_type="standard_deviation",
        breakpoint_threshold_amount=1
        
)
```

```python
sample = """
The quick brown fox jumps over the lazy dog. The lazy dog was not happy about this. The quick brown fox was very happy about this. The lazy dog decided to take a nap instead of chasing the quick brown fox. The quick brown fox continued to jump over the lazy dog while it was napping. The lazy dog woke up and was very angry about this. The quick brown fox apologized and promised to stop jumping over the lazy dog. The lazy dog accepted the apology and they became friends. The quick brown fox jumps over the lazy dog. The lazy dog was not happy about this. The quick brown fox was very happy about this. The lazy dog decided to take a nap instead of chasing the quick brown fox. The quick brown fox continued to jump over the lazy dog while it was napping. The lazy dog woke up and was very angry about this. The quick brown fox apologized and promised to stop jumping over the lazy dog. The lazy dog accepted the apology and they became friends.
"""
```

```python
docs = semantic_text_splitter.create_documents([sample])

print(len(docs))
print(docs)
```
